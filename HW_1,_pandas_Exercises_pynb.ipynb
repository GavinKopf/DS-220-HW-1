## Working with Tabular Data: Pandas Exercises with Gapminder
This dataset is the famous [Gapminder dataset](https://www.gapminder.org/data/), containing historical data on life expectancy, population, and GDP per capita for various countries.

Use the pandas skills you learned in *Intro to Data Science* to answer the following descriptive statistics questions. Below each question, write some Python that determines the answer.

Store the value of your solution in the variable 'solution_<number>', eg **```solution_1 = 42```**

####1. How many distinct entries (unique countries) are there in the `country` column?

````python
solution_1 = gapminder_data["country"].nunique()
solution_1
````

#### Unit Test for Problem 1

````python
assert solution_1 == 142, f"Expected 142 unique countries, got {solution_1}"
print("Unit test for Problem 1 passed!")
````

####2. What is the most frequent entry in the `continent` column?

````python
solution_2 = gapminder_data["continent"].mode()[0]
solution_2
````

#### Unit Test for Problem 2

````python
assert solution_2 == 'Africa', f"Expected 'Africa' as most frequent continent, got {solution_2}"
print("Unit test for Problem 2 passed!")
````

####3. How many records in the dataset are from the year `2007`?

````python
solution_3 = int((gapminder_data["year"] == 2007).sum())
solution_3
````

#### Unit Test for Problem 3

````python
assert solution_3 == 142, f"Expected 142 records for year 2007, got {solution_3}"
print("Unit test for Problem 3 passed!")
````

####4. Create a sub-DataFrame containing only the records where the continent is `Asia`, and only the columns `country`, `year`, and `lifeExp`.

````python
solution_4 = gapminder_data.loc[gapminder_data["continent"] == "Asia", ["country", "year", "lifeExp"]]
solution_4
````

#### Unit Test for Problem 4

````python
import pandas as pd
assert isinstance(solution_4, pd.DataFrame), "Solution should be a pandas DataFrame"
assert solution_4.shape == (396, 3), f"Expected shape (396, 3), got {solution_4.shape}"
expected_columns = ['country', 'year', 'lifeExp']
assert list(solution_4.columns) == expected_columns, f"Expected columns {expected_columns}, got {list(solution_4.columns)}"
print("Unit test for Problem 4 passed!")
````

####5. What was the highest life expectancy (`lifeExp`) achieved by a country in `Africa`?

````python
solution_5 = gapminder_data.loc[gapminder_data['continent'] == 'Africa', ['lifeExp']].max()
solution_5 = float(solution_5.iloc[0])
solution_5
````

#### Unit Test for Problem 5

````python
assert abs(solution_5 - 76.442) < 0.001, f"Expected highest life expectancy in Africa around 76.442, got {solution_5}"
print("Unit test for Problem 5 passed!")
````

####6. Which country in `Europe` had the highest GDP per capita (`gdpPercap`)? Your result should be the **name** of the country.

````python
solution_6 = gapminder_data.loc[gapminder_data['continent'] == 'Europe', ['country','gdpPercap']].sort_values(by='gdpPercap', ascending= False).iloc[0]['country']
solution_6
````

#### Unit Test for Problem 6

````python
assert solution_6 == 'Norway', f"Expected 'Norway' as country with highest GDP per capita in Europe, got {solution_6}"
print("Unit test for Problem 6 passed!")
````

####7. Let's now combine several of these data tasks...

You are an economist studying evolution of continental wealth trends. You want to identify the year of maximum difference between GDP per capita in
the Americas and Asia.

In other words, ***which year had the largest gap between average GDP in
the Americas versus Asia***?

Use Python, pandas, and Gapminder dataset to answer this question.

````python
am = gapminder_data.loc[gapminder_data['continent'] == 'Americas', ['country','year','gdpPercap']].groupby('year')["gdpPercap"].mean()
asia = gapminder_data.loc[gapminder_data['continent'] == 'Asia', ['country','year','gdpPercap']].groupby('year')["gdpPercap"].mean()

solution_7 = (asia - am).abs()
solution_7
````

#### Unit Test for Problem 7

````python
assert solution_7 == 1987, f"Expected 1987 as the year with the largest positive difference in average GDP per capita between Americas and Asia, got {solution_7}"
print("Unit test for Problem 7 passed!")
````
