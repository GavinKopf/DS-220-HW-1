### A. ReLU Function
Using a conditional, write the [relu](https://en.wikipedia.org/wiki/Rectifier_(neural_networks)) function defined as follows:

$$\text{relu}(x) = \begin{cases} x, & \text{if } x \ge 0 \\ 0, & \text{otherwise } \end{cases}$$

```python
def relu(x):
    if x >= 0:
        return x
    else:
        return 0

# Execution
print(f'relu(5) = {relu(5)}, relu(-5) = {relu(-5)}')

# Tests
run_test(relu(5) == 5, "ReLU positive")
run_test(relu(-5) == 0, "ReLU negative")
run_test(relu(0) == 0, "ReLU zero")
```

### B. Euclidean Norm
Write a function that computes the **Euclidean norm** of a vector, represented as a list. The Euclidean norm is the square root of the sum of squares:

$$sqrt(\sum_{i=1}^{k} l_i^2)$$

```python
import math

def norm(l):
    sum_sq = 0
    for x in l:
        sum_sq += x**2
    return math.sqrt(sum_sq)

# Execution
print(f'norm([3, 4]) = {norm([3, 4])}')

# Tests
run_test(norm([3, 4]) == 5.0, "Euclidean norm [3, 4]")
run_test(norm([0, 0]) == 0.0, "Euclidean norm [0, 0]")
run_test(norm([1]) == 1.0, "Euclidean norm [1]")
```

### C. List Reverse
Write the function `list_reverse` that takes a list and returns a reversed version of the list.

**Constraint:** Define the function without calling the built-in `.reverse()` method or using slicing like `x[::-1]`.

```python
def list_reverse(l):
    reversed_list = []
    for i in range(len(l) - 1, -1, -1):
        reversed_list.append(l[i])
    return reversed_list

# Execution
print(f'list_reverse([1, 2, 3]) = {list_reverse([1, 2, 3])}')

# Tests
run_test(list_reverse([1, 2, 3]) == [3, 2, 1], "List reverse standard")
run_test(list_reverse([]) == [], "List reverse empty")
```

### D. Shakespeare Word Count
Write a function to count the number of distinct **4-letter words** in the Shakespeare file available at:
`http://composingprograms.com/shakespeare.txt`.

Hint: Look at [Composing Programs deck ](https://colab.research.google.com/drive/1CE-b3OLXC7S4Qigbf_SzPwe9eUKuAV-6#scrollTo=2sUycc4n8mNb&line=2&uniqifier=1)for similar example.


```python
from urllib.request import urlopen

def count_four() -> set:
    url = "http://composingprograms.com/shakespeare.txt"
    text = urlopen(url).read().decode().lower()
    words = set()
    current_word = []
    for char in text:
        if 'a' <= char <= 'z':
            current_word.append(char)
        else:
            if len(current_word) == 4:
                words.add("".join(current_word))
            current_word = []
    # Check for a word at the very end of the text
    if len(current_word) == 4:
        words.add("".join(current_word))
    return words

# Tests
four_letter_words = count_four()
run_test(len(four_letter_words) > 0, "Shakespeare 4-letter words found")
run_test(all(len(w) == 4 for w in four_letter_words), "All words in set are length 4")
```

### E. Combine Dictionaries
Write a function to take two dictionaries as parameters and return a combined version. If the dictionaries share a key, its value should be the **sum** of the two values.

```python
def combine(d1, d2):
    combined_dict = d1.copy()
    for key, value in d2.items():
        if key in combined_dict:
            combined_dict[key] += value
        else:
            combined_dict[key] = value
    return combined_dict

# Execution
d1 = {'key1': 50, 'key2': 100}
d2 = {'key1': 200, 'key3': 300}
combined = combine(d1, d2)
print(f'Combined dictionary: {combined}')

# Tests
run_test(combined.get('key1') == 250, "Combine dict shared key")
run_test(combined.get('key3') == 300, "Combine dict unique key")
```

### F. Maximum of a List
Using a `for` loop and a conditional, write a function `max_list` that returns the maximum value in a vector.

**Constraint:** Do not use the built-in `max()` function.

```python
def max_list(l):
    if not l:
        return None
    max_val = l[0]
    for x in l:
        if x > max_val:
            max_val = x
    return max_val

# Execution
print(f'max_list([2, 5, 1, 9]) = {max_list([2, 5, 1, 9])}')

# Tests
run_test(max_list([2, 5, 1, 9]) == 9, "Max list standard")
run_test(max_list([-1, -5, -2]) == -1, "Max list negative numbers")
run_test(max_list([]) is None, "Max list empty")
```

### G. Prime Testing
Define a function `is_prime` to test if a number is prime. Run it and show the return values for the numbers: 2, 4, 7, 17, -3, 1.

```python
import math

def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    # Check for factors from 3 up to sqrt(n) (inclusive) with a step of 2
    for i in range(3, int(math.sqrt(n)) + 1, 2):
        if n % i == 0:
            return False
    return True

# Execution
test_nums = [2, 4, 7, 17, -3, 1]
primes = [n for n in test_nums if is_prime(n)]
print(f'Prime numbers in {test_nums}: {primes}')

# Tests
run_test(is_prime(2) == True, "Prime 2")
run_test(is_prime(4) == False, "Prime 4")
run_test(is_prime(1) == False, "Prime 1")
run_test(is_prime(-3) == False, "Prime negative")
run_test(is_prime(7) == True, "Prime 7")
run_test(is_prime(17) == True, "Prime 17")
```

### H & I. Common Elements and Removing Duplicates
1.  **Common Elements:** Find elements shared between two lists without using sets.
2.  **Remove Duplicates:** Keep only the first occurrence of each number in a list without using sets.

```python
def common_elements(l1, l2):
    result = []
    for item1 in l1:
        if item1 in l2 and item1 not in result:
            result.append(item1)
    return result

def remove_duplicates(l):
    result = []
    for item in l:
        if item not in result:
            result.append(item)
    return result

# Execution and Tests
l1 = [i for i in range(0, 100, 2)]
l2 = [i for i in range(0, 100, 3)]

res_common = common_elements(l1, l2)
res_nodup = remove_duplicates(l1 + l2)

print(f"Common elements: {res_common}")
print(f"Combined without duplicates length: {len(res_nodup)}")

# Tests
run_test(res_common[0:3] == [0, 6, 12], "Common elements start sequence check")
run_test(len(res_common) == 17, "Common elements total count check")
run_test(len(res_nodup) == 67, "Remove duplicates combined length check")
```

### J. Apply Function N Times
Write a function `apply_n_times(f, arg, n)` that calculates $f(f(...f(arg)))$.

**Short-circuit:** If $f(x) = x$, return $x$ immediately.

Use the functions plus_one and plus_one_lt_30 to help you test.

```python
def apply_n_times(f, arg, n):
    current_val = arg
    for _ in range(n):
        next_val = f(current_val)
        if next_val == current_val:
            return current_val
        current_val = next_val
    return current_val

def plus_one(n): return n + 1

# Helper for short-circuit testing
def plus_one_lt_30(n):
    if n < 30:
        return n + 1
    return n

# Execution
print(f'apply_n_times(plus_one, 0, 5) = {apply_n_times(plus_one, 0, 5)}')
print(f'apply_n_times(plus_one_lt_30, 28, 5) = {apply_n_times(plus_one_lt_30, 28, 5)}')

# Tests
run_test(apply_n_times(plus_one, 0, 5) == 5, "Apply n times standard")
run_test(apply_n_times(plus_one_lt_30, 28, 5) == 30, "Apply n times short-circuit at limit")
run_test(apply_n_times(lambda x: x, 10, 5) == 10, "Apply n times identity short-circuit")
```

### A. Normalize List to Nested Dictionary
Define a function `normalize_list` that takes a list of lists (containing book name, chapter name, pages, and people) and creates a hierarchical dictionary where the outer key is the book name and the inner key is the chapter name.

```python
def normalize_list(l: list[list]) -> dict:
    normalized_dict = {}
    for item in l:
        book_name, chapter_name, pages, people = item
        if book_name not in normalized_dict:
            normalized_dict[book_name] = {}
        normalized_dict[book_name][chapter_name] = (pages, people)
    return normalized_dict

# Execution
L = [['Tale of Two Cities', 'Recalled To Life', 233, 43],
     ['Tale of Two Cities', 'The Golden Thread', 137, 72],
     ['Oliver Twist', 'Life in a Workhouse', 39, 7],
     ['Oliver Twist', 'A New Home', 59, 17]]

D_actual = normalize_list(L)
print(f"Normalized Dictionary: {D_actual}")

# Tests
run_test('Tale of Two Cities' in D_actual, "Normalize list outer key check")
run_test(len(D_actual) == 2, "Normalize list number of books check")
run_test(D_actual['Oliver Twist']['A New Home'] == (59, 17), "Normalize list content verification")
```

### B. Denormalize Dictionary to List
Define a function `denormalize_dict` that performs the reverse operation: it takes the nested dictionary and flattens it back into a list of lists.

```python
def denormalize_dict(d: dict) -> list[list]:
    result_list = []
    for book_name, chapters in d.items():
        for chapter_name, (pages, people) in chapters.items():
            result_list.append([book_name, chapter_name, pages, people])
    return result_list

# Tests
L_back = denormalize_dict(D_actual)
print(f"Denormalized List: {L_back}")


# Unit Tests
run_test(isinstance(L_back, list), "Denormalize return type is list")
run_test(len(L_back) == 4, "Denormalize list length check (4 entries)")
run_test(['Oliver Twist', 'Life in a Workhouse', 39, 7] in L_back, "Denormalize content verification")
```

### A. Planet Class and Data Processing
1.  Create a class `Planet` for the planets data.
2.  Include methods to calculate the planet's volume and density.
3.  Create a list of `Planet` objects using the provided data.
4.  Create a dictionary mapping planet names to their radius, volume, and density.
5.  Print the planets in order of increasing density.

```python
import math

class Planet:
    def __init__(self, name, mass, radius, distance_from_the_sun):
        self.name = name
        self.mass = mass
        self.radius = radius
        self.distance_from_the_sun = distance_from_the_sun

    def calculate_volume(self):
        return (4/3) * math.pi * (self.radius**3)

    def calculate_density(self):
        volume = self.calculate_volume()
        if volume == 0:
            return 0  # Avoid division by zero
        return self.mass / volume

# Planet Data
planet_data = {
    'Mercury': (3.18e23, 2.43e6, 5.79e10),
    'Venus': (4.88e24, 6.06e6, 1.08e11),
    'Earth': (5.98e24, 6.38e6, 1.496e11),
    'Mars': (6.42e23, 3.37e6, 2.28e11),
    'Jupiter': (1.90e27, 6.99e7, 7.78e11),
    'Saturn': (5.68e26, 5.85e7, 1.43e12),
    'Uranus': (8.68e25, 2.33e7, 2.87e12),
    'Neptune': (1.03e26, 2.21e7, 4.50e12)
}

# Create a list of Planet objects
planets = []
planet_properties = {}

for name, (mass, radius, distance) in planet_data.items():
    planet = Planet(name, mass, radius, distance)
    planets.append(planet)
    planet_properties[name] = {
        'radius': planet.radius,
        'volume': planet.calculate_volume(),
        'density': planet.calculate_density()
    }

# Print planets in order of increasing density
print("Planets ordered by increasing density:")
sorted_planets = sorted(planets, key=lambda p: p.calculate_density())
for planet in sorted_planets:
    print(f"  {planet.name}: Density = {planet.calculate_density():.3e} kg/m^3")

# Tests
earth_data = planet_data['Earth']
earth_p = Planet('Earth', *earth_data)
run_test(round(earth_p.calculate_volume(), -18) == 1.088e21, "Earth volume calculation (planet_data)")
run_test(len(planets) == 8, "Planet objects creation count")
```
